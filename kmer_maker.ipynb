```python
import os

kmer_size = 4

def kmer_maker(kmer_size, dna):
    kmer = []
    for start in range(0, len(dna) - kmer_size + 1):
        seq = dna[start:start + kmer_size]
        kmer.append(seq)
    return kmer

# Look at files ending in .dna in the current folder, one by one
for file_name in os.listdir('.'):
    if file_name.endswith('.dna'):
        print('Found a file named:', file_name)

        with open(file_name) as file_object:
            for line in file_object:
                dna = line.rstrip('\n')
                if not dna:          # skip empty lines
                    continue
                print('Found a DNA sequence with length of', len(dna))

                kmers = kmer_maker(kmer_size, dna)
                print('Number of', str(kmer_size) + '-mers:', len(kmers))
                print(kmers)
```
