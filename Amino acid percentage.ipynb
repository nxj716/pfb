```python
#calculate AA percentage in a protein sequence.

standard_aa = [
    'A', 'C', 'D', 'E', 'F',
    'G', 'H', 'I', 'K', 'L',
    'M', 'N', 'P', 'Q', 'R',
    'S', 'T', 'V', 'W', 'Y'
]

def get_aa_p(protein):
  results = {}
  for r in standard_aa:
    aa_p = (protein.count(r) / len(protein)) * 100
    results[r] = round(aa_p, 2)
    #results is a dictionary. r is key, = is to assign a value.
  return results


get_aa_p('MSRSLLLRFLLFLLLLPPLP')
```
